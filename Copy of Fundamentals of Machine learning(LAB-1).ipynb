```python
import pandas as pd

def find_s_algorithm(data):
    concepts = data.iloc[:, :-1].values
    target = data.iloc[:, -1].values

    num_attributes = concepts.shape[1]
    hypothesis = ['0'] * num_attributes

    print("Initial Hypothesis:", hypothesis)
    print("-" * 50)

    for i, example in enumerate(concepts):

        if target[i] == 'Yes':
            for j in range(num_attributes):
                if hypothesis[j] == '0':
                    hypothesis[j] = example[j]
                elif hypothesis[j] != example[j]:
                    hypothesis[j] = '?'
            print(f"Sample {i+1} ({target[i]}): {example}")
            print(f"Updated Hypothesis: {hypothesis}\n")
        else:
            print(f"Sample {i+1} ({target[i]}): {example}")
            print(f"Hypothesis unchanged: {hypothesis}\n")

    return hypothesis


data_dict = {
    'Temp': ['Hot', 'Hot', 'Iced'],
    'Milk': ['Oat', 'Whole', 'Oat'],
    'Flavor': ['Vanilla', 'Hazelnut', 'Vanilla'],
    'Enjoyed': ['Yes', 'No', 'Yes']
}

df = pd.DataFrame(data_dict)

print("=== Training Dataset ===")
print(df)
print("\n=== FIND-S Algorithm Trace ===")
final_hypothesis = find_s_algorithm(df)

print("=" * 50)
print("Final Most Specific Hypothesis:", final_hypothesis)
```
